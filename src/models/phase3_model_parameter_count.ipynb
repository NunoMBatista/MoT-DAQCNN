```python
import torch
from phase3_vit_comparison import Phase3ExperimentModel

def count_parameters(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

# Configuration for counts
configs = {
    "DeiT Classical": {"is_hybrid": False, "from_scratch": False, "name": "deit_tiny_patch16_224"},
    "DeiT Hybrid":    {"is_hybrid": True,  "from_scratch": False, "name": "deit_tiny_patch16_224"},
    "Scratch Classical": {"is_hybrid": False, "from_scratch": True, "num_layers": 5, "embed_dim": 192},
    "Scratch Hybrid":    {"is_hybrid": True,  "from_scratch": True, "num_layers": 5, "embed_dim": 192}
}

print(f"{'Model Type':<25} | {'Total Trainable Params':<25}")
print("-" * 55)

for label, args in configs.items():
    model = Phase3ExperimentModel(
        is_hybrid=args['is_hybrid'],
        from_scratch=args['from_scratch'],
        model_name=args.get('name', "deit_tiny_patch16_224"),
        num_layers=args.get('num_layers', 12),
        embed_dim=args.get('embed_dim', 192),
        q_channels=180
    )
    
    params = count_parameters(model)
    print(f"{label:<25} | {params:>25,}")
```
